```python
import io
import urllib.request
import urllib.parse
import zipfile
import pvlib
import pandas as pd

class WeatherDownloader:
    """Downloads official EnergyPlus weather files and handles strict host validation."""

    ENERGYPLUS_WEATHER_HOST = 'energyplus-weather.s3.amazonaws.com'

    def __init__(self, timeout: int = 30):
        self.timeout = timeout

    def download_and_extract(self, zip_file_url: str) -> str:
        clean_url = zip_file_url.strip()
        parsed_url = urllib.parse.urlparse(clean_url)
        hostname = (parsed_url.hostname or '').lower()

        if (parsed_url.scheme != 'https' or
            self.ENERGYPLUS_WEATHER_HOST not in hostname or
            not parsed_url.path.lower().endswith('.zip')):
            raise ValueError(f"Invalid host! The weather_url must belong to {self.ENERGYPLUS_WEATHER_HOST}")

        with urllib.request.urlopen(clean_url, timeout=self.timeout) as response:
            archive_bytes = response.read()

        with zipfile.ZipFile(io.BytesIO(archive_bytes)) as archive:
            epw_members = [
                info for info in archive.infolist()
                if not info.is_dir() and info.filename.lower().endswith('.epw')
            ]
            if len(epw_members) != 1:
                raise ValueError("Archive format error: Expected exactly one EPW file inside the ZIP archive.")

            return archive.read(epw_members[0]).decode('utf-8')

class WeatherReader:
    """Parses raw extracted weather strings into clean tabular datasets using pvlib."""

    @staticmethod
    def parse_data(epw_string_data: str):
        with io.StringIO(epw_string_data) as epw_file:
            data, metadata = pvlib.iotools.read_epw(epw_file)
        return data, metadata

if __name__ == "__main__":
    # Clean official URL (without markdown asterisks)
    target_url = (
        "https://energyplus-weather.s3.amazonaws.com/"
        "north_and_central_america_wmo_region_4/USA/CA/"
        "USA_CA_San.Francisco.Intl.AP.724940_TMY3/"
        "USA_CA_San.Francisco.Intl.AP.724940_TMY3.zip"
    )

    try:
        downloader = WeatherDownloader(timeout=30)
        raw_epw_content = downloader.download_and_extract(target_url)

        data, metadata = WeatherReader.parse_data(raw_epw_content)

        # Safe dictionary lookups using keys present in pvlib metadata
        city_name = metadata.get('city', metadata.get('name', 'Unknown Location'))

        print("Verification Details:")
        print(f"Location Verified: {city_name}")
        print(f"Coordinates: Lat {metadata.get('latitude')}, Lon {metadata.get('longitude')}")
        print(f"Timezone: {metadata.get('tz')}")

        print("\nParsed Data Sample (Pandas DataFrame View):")
        display(data[['temp_air', 'relative_humidity', 'wind_speed', 'ghi']].head())

    except Exception as e:
        print(f"Execution Failure: {e}")
```
